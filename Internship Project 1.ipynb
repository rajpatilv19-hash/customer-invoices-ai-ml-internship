```python
%%writefile clean_invoices.py

import pandas as pd
import os

# File paths
input_file = r"C:\Users\rajpa\train_invoices.csv"
output_file = r"C:\Users\rajpa\cleaned_customer_invoices.csv"
log_file = r"C:\Users\rajpa\customer_invoices_cleaning_log.csv"

print("Starting invoice data cleaning...")
```

```python

with open("clean_invoices.py", "a", encoding="utf-8") as f:
    f.write("""
# Load original dataset
print("Loading original dataset...")

df = pd.read_csv(input_file)

print("Dataset loaded successfully!")
print("Original dataset shape:", df.shape)

# Standardize column names
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Remove leading and trailing spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Column names and text values standardized.")
""")

print("Data-loading code added to clean_invoices.py")
```

```python

with open("clean_invoices.py", "a", encoding="utf-8") as f:
    f.write("""
# Standardize categorical columns
df["payment_status"] = df["payment_status"].str.title()

df["currency"] = df["currency"].str.upper()

category_columns = [
    "payment_method",
    "customer_country",
    "product_category",
    "order_channel"
]

for col in category_columns:
    df[col] = df[col].str.title()

# Correct UPI capitalization
df["payment_method"] = df["payment_method"].replace({
    "Upi": "UPI"
})

print("Categorical values standardized.")

# Convert date columns to datetime
date_columns = [
    "invoice_date",
    "due_date",
    "payment_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("Date columns converted to datetime.")

# Preserve original payment dates
df["original_payment_date"] = df["payment_date"].copy()

# Keep confirmed payment dates only for Paid invoices
paid_mask = df["payment_status"] == "Paid"

df["payment_date"] = df["payment_date"].where(paid_mask)

print("Original payment dates preserved.")
print("Non-paid payment dates masked.")
""")

print("Date and category cleaning code added!")
```

```python

with open("clean_invoices.py", "a", encoding="utf-8") as f:
    f.write("""
# Feature Engineering

# 1. Calculate days taken to pay
df["days_to_pay"] = (
    df["payment_date"] - df["invoice_date"]
).dt.days

# 2. Calculate payment delay
df["delay"] = (
    df["payment_date"] - df["due_date"]
).dt.days

# 3. Identify late payments
df["paid_late"] = pd.Series(
    pd.NA, index=df.index, dtype="boolean"
)

paid_mask = df["payment_status"] == "Paid"

df.loc[paid_mask, "paid_late"] = (
    df.loc[paid_mask, "delay"] > 0
)

# 4. Classify invoice state
df["open_paid_flag"] = df["payment_status"].map({
    "Paid": "Paid",
    "Pending": "Open",
    "Overdue": "Open",
    "Cancelled": "Cancelled"
})

print("Feature engineering completed!")
""")

print("Feature engineering code added!")
```

```python

import pandas as pd
import os
import shutil

# Create project folder
project_folder = r"C:\Users\rajpa\Internship Project 1"
os.makedirs(project_folder, exist_ok=True)

# File paths
cleaned_file = r"C:\Users\rajpa\cleaned_customer_invoices_from_script.csv"
script_file = r"C:\Users\rajpa\clean_invoices.py"
log_file = r"C:\Users\rajpa\customer_invoices_cleaning_log.csv"

# Create detailed cleaning log
log_data = [
    ["Missing values in original data", "Checked all columns", "No missing values found"],
    ["Duplicate invoice IDs", "Checked invoice_id", "0 duplicates"],
    ["Duplicate invoice numbers", "Checked invoice_number", "0 duplicates"],
    ["Invalid invoice dates", "Parsed dates using errors='coerce'", "0 invalid dates"],
    ["Due date before invoice date", "Compared due_date and invoice_date", "0 invalid records"],
    ["Negative or zero invoice amounts", "Checked total_amount", "0 invalid amounts"],
    ["Invalid numerical values", "Checked quantity, price, discount, tax and shipping", "0 invalid records"],
    ["Text standardization", "Removed extra spaces and standardized text", "Completed"],
    ["Payment method standardization", "Corrected Upi to UPI", "Completed"],
    ["Non-paid payment dates", "Preserved original dates and masked dates for non-Paid invoices", "799965 dates masked; mentor confirmation pending"],
    ["Days to pay and delay", "Created days_to_pay and delay", "Created for Paid invoices"],
    ["Paid late indicator", "Created paid_late using delay > 0", "0 paid-late invoices found"],
    ["Invoice state", "Created open_paid_flag", "Paid, Open and Cancelled categories created"]
]

cleaning_log = pd.DataFrame(
    log_data,
    columns=["Issue", "Action Taken", "Result"]
)

# Save log inside project folder
project_log = os.path.join(project_folder, "customer_invoices_cleaning_log.csv")
cleaning_log.to_csv(project_log, index=False)

# Move cleaned dataset into project folder
project_cleaned = os.path.join(
    project_folder, "cleaned_customer_invoices.csv"
)

if os.path.exists(cleaned_file):
    if not os.path.exists(project_cleaned):
        shutil.move(cleaned_file, project_cleaned)
    else:
        print("Cleaned dataset already exists in project folder.")

# Copy reusable script into project folder
project_script = os.path.join(project_folder, "clean_invoices.py")

if os.path.exists(script_file):
    shutil.copy2(script_file, project_script)

# Update paths in the project copy of the script
if os.path.exists(project_script):
    with open(project_script, "r", encoding="utf-8") as f:
        script_text = f.read()

    script_text = script_text.replace(
        r"C:\Users\rajpa\train_invoices.csv",
        r"C:\Users\rajpa\train_invoices.csv"
    )
    script_text = script_text.replace(
        r"C:\Users\rajpa\cleaned_customer_invoices_from_script.csv",
        project_cleaned
    )
    script_text = script_text.replace(
        r"C:\Users\rajpa\customer_invoices_cleaning_log.csv",
        project_log
    )

    with open(project_script, "w", encoding="utf-8") as f:
        f.write(script_text)

print("Project folder created and organized!")
print("Cleaning log entries:", len(cleaning_log))
print("Project folder:", project_folder)
print("\nFiles in project folder:")
for file in os.listdir(project_folder):
    print("-", file)
```

```python

import pandas as pd

folder = r"C:\Users\rajpa\Internship Project 1"

# Update cleaning log
log_file = folder + r"\customer_invoices_cleaning_log.csv"

log = pd.read_csv(log_file)

log.loc[
    log["Issue"] == "Non-paid payment dates",
    "Result"
] = "Confirmed: payment_date is blank for non-Paid invoices; original dates preserved."

log.to_csv(log_file, index=False)

# Verify payment dates in cleaned dataset
cleaned_file = folder + r"\cleaned_customer_invoices.csv"

non_paid_with_date = 0
paid_without_date = 0

for chunk in pd.read_csv(
    cleaned_file,
    chunksize=100000,
    usecols=["payment_status", "payment_date"]
):
    non_paid_with_date += (
        (chunk["payment_status"] != "Paid") &
        chunk["payment_date"].notna()
    ).sum()

    paid_without_date += (
        (chunk["payment_status"] == "Paid") &
        chunk["payment_date"].isna()
    ).sum()

print("Non-paid invoices with payment dates:", non_paid_with_date)
print("Paid invoices without payment dates:", paid_without_date)
print("Cleaning log updated successfully!")
```

```python

import pandas as pd

file_path = r"C:\Users\rajpa\Internship Project 1\cleaned_customer_invoices.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()

```

```python

# Standardize country names
df["customer_country"] = df["customer_country"].replace({
    "Uk": "UK",
    "Usa": "USA",
    "Uae": "UAE"
})

# Convert date columns into datetime format
date_columns = [
    "invoice_date",
    "due_date",
    "payment_date",
    "original_payment_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("Country standardization completed!")
print("\nCountry values:")
print(sorted(df["customer_country"].unique()))

print("\nDate data types:")
print(df[date_columns].dtypes)

```

```python

# Invoice amount statistics by currency

amount_summary = df.groupby("currency")["total_amount"].agg(
    ["count", "mean", "median", "min", "max", "sum"]
).round(2)

print("Invoice Amount Analysis by Currency:")
display(amount_summary)
```
